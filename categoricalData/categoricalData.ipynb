```python
import pandas as pd

df = pd.DataFrame({
    'Department': ['HR', 'HR', 'IT', 'IT'],
    'Level': ['Junior', 'Senior', 'Junior', 'Senior'],
    'Salary': [50000, 70000, 60000, 80000]
})
```

```python
print(df.dtypes)
```

```python
# converting to catgorical data type
df['Department'] = df['Department'].astype('category')
df['Level'] = df['Level'].astype('category')
print(df.dtypes)
```

```python
df['Level'] = df['Level'].cat.rename_categories({
    'Junior':'Jr',
    'Senior':'Sr'
})
print(df)
```

```python
# category codes
df['department_code'] = df['Department'].cat.codes
print(df)
```
