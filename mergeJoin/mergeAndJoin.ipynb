### one to one join

```python
import pandas as pd

df1 = pd.DataFrame({
    "employee": ["Alice", "Bob", "Charlie", "David"],
    "department": ["HR", "IT", "Finance", "Finance"],
    "address": ["123 Main St", "456 Elm St", "789 Oak St", "101 Pine St"]
})

df2 = pd.DataFrame({
    "employee": ["Alice", "Bob", "Charlie"],
    "skills": [["arcitecture", "design"], ["development", "testing"], ["analysis"]],
    "salary": [50000, 60000, 70000]
})
```

```python
df3 = pd.merge(df1,df2) # default is inner join, on the common column "employee"
df3
```

### many to one join

```python
df4 = pd.DataFrame({
    "department": ["HR", "IT", "Finance"],
    "supervisor": ["Eve", "Frank", "Grace"]
})

df5 = pd.merge(df1,df4) # default is inner join, on the common column "employee"
df5
```

```python
df_left_merged = pd.merge(df1, df2, on="employee", how="left")
df_left_merged
```

```python
df_right_merged = pd.merge(df1, df2, on="employee", how="right")
df_right_merged
```

```python
df_outer_merged = pd.merge(df1, df2, on="employee", how="outer")
df_outer_merged
```
