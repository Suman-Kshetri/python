## Collections Module
● Python has a built-in module called collections. It gives you extra data structures that are more powerful than lists and dicts.
1) Counter
2) defaultdict
3) deque
4) namedtuple

### Counter
-> counts how many times each item appears

```python
from collections import Counter

nums = [1,2,4,5,5,4,4,2,1]
c = Counter(nums)
print(c)
print(type(c))
```

## defaultdict
-> create default values automatically

```python
from collections import defaultdict
d = defaultdict(int)
print(d)
d['a'] +=1
print(d)
```

## deque
● Fast for adding/removing from both sides.

```python
from collections import deque

q= deque([1,2,4,5])
q.appendleft(0)
q.append(3)
print(q)
```

## namedtuple
● Works like a lightweight object.

```python
from collections import namedtuple

point = namedtuple('point', 'x y')
p = point(3,4)
print(p.x,p.y)
```

## Iterators
● An iterator lets you loop through items one by one.
● It remembers where it is, so it uses less memory.

```python
nums = [10,30,20]
it = iter(nums)
print(next(it))
print(next(it))
```

```python
nums = [10, 20 ,30, 54, 39, 89]
it = iter(nums)

while True:
    try:
        x = next(it)
        print(x)
    except StopIteration:
        break
```

## Generators
● A generator is a simple way to build an iterator with yield.
● It saves memory because it produces values only when needed

## Decorators
● A decorator adds extra behavior to a function without changing it.

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print("time: ", end - start)
        return result
    return wrapper

@timer
def work():
    for _ in range(100000):
        pass

work()
```

## Lambda Expressions
● A lambda is a small, anonymous function.
● You use it when you need a simple function for a short time.

**lambda arguments: expression**

```python
add = lambda a,b:a+b
print(add(2,3))
```

# 1. Sorting

```python
students = [("Ram",25),("Hari", 20)]
student_sorted = sorted(students, key=lambda x: x[1])
student_sorted
```

# 2. Filtering

```python
nums = [1,3,4,5,6]
evens = list(filter(lambda x : x % 2 == 0, nums))
evens
```

# 3. Mapping

```python
nums = range(1,10)
squares = list(map(lambda x:x * x, nums))
squares
```

# OOPs

```python
class Complex:
    def __init__ (self, real, img):
        self.real  = real
        self.img = img
    def __str__ (self):
        return f"{self.real} + {self.img} i"
    def __add__ (self, other):
        r = self.real + other.real
        i = self.img + other.img
        return Complex(r,i)
```

```python
a = Complex(3,4)
b = Complex(5,6)

print(a)
print(b)

c = a + b
print(c)
```
