```python
import pandas as pd
```

```python
df = pd.read_csv("students_missing.csv")
df.head()
```

```python
print(df.isna().sum())
```

```python
# Drop rows with missing values
df_dropped = df.dropna()
print(df_dropped)
```

```python
# Drop columns with missing values
df_dropped_columns = df.dropna(axis=1)
print(df_dropped_columns)
```

```python
# fill missing values with a specific value constant

df_filled_constant = df.fillna(0)
print(df_filled_constant)
```

```python
# fill missing values with mean
# Fill numeric columns
df["Math"].fillna(df["Math"].mean(), inplace=True)
df["Science"].fillna(df["Science"].mean(), inplace=True)
df["English"].fillna(df["English"].mean(), inplace=True)

# Fill categorical column
df["Name"].fillna("Unknown", inplace=True)
print(df)
```

```python
# forward fill
df_forward_filled = df.ffill()
print(df_forward_filled)
```

```python
# backward fill
df_backward_filled = df.bfill()
print(df_backward_filled)
```
