### wide data fromat
- Each variable is a seperate column
- Easy to read and understand

```python
import pandas as pd

df_wide = pd.DataFrame({
    "student": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Math": [85, 90, 95, 88, 92],
    "Science": [80, 85, 90, 82, 87],
    "English": [88, 92, 94, 86, 90]
})
print(df_wide)
```

### melt => converts wide data to long format
- converts columns into rows
- useful for data analysis and visualization

```python
df_melt = pd.melt(
    df_wide,
    id_vars=["student"],
    var_name="subject",
    value_name="score"
)
print(df_melt)
```

### pivot => converts long data to wide format
- converts rows into columns
- useful for data analysis and visualization

```python
df_pivot = df_melt.pivot(index="student", columns="subject", values="score")
print(df_pivot)
```

### stack => combines multiple columns into a single column
- Result is a Series with MultiIndex.

```python
stacked = df_wide.stack()
print(stacked)
```

```python
# unstacked -> reverses the stacking operation
unstacked = stacked.unstack()
print(unstacked)
```

### pivot table
- Pivot table is an advanced version of pivot.
- It supports aggregation.
- Pivot table handles duplicates using aggregation.

```python
df_exam = pd.DataFrame({
    "student": ["Alice", "Bob", "Charlie", "David", "Eve", "Alice", "Bob", "Charlie", "David", "Eve"],
    "subject": ["Math", "Math", "Math", "Math", "Math", "Science", "Science", "Science", "Science", "Science"],
    "score": [85, 90, 95, 88, 92, 80, 85, 90, 82, 87]
})

pivot_table = df_exam.pivot_table(
    index="student",
    columns="subject",
    values="score",
    aggfunc="mean"
)
print(pivot_table)
```
