```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0, 10, 0.1)
y = np.square(x)

plt.plot(x, y)
plt.title("Square Function")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.show()
```

```python
# vertical bar plot
categories = ["Apples", "Bananas", "Cherries", "Chocolate"]
values = [25, 30, 15, 20]

plt.bar(categories, values)
plt.title("Fruit Sales")
plt.xlabel("Fruits")
plt.ylabel("Sales")
plt.show()
```

```python
colors = ["red", "yellow", "pink", "brown"]
plt.barh(categories, values, color=colors)
plt.title("Fruit Sales")
plt.xlabel("Sales")
plt.ylabel("Fruits")
plt.show()
```

```python
# histrogram plot

# Generate random samples
import numpy as np

data = np.random.normal(loc=50, scale=10, size=1000)

plt.hist(data, bins=30, color="blue", edgecolor="black", alpha=0.7)
plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid()
plt.show()
```

```python
# scatter plot > shows relationship between two variables
# > detect correlation, clusters, outliers

from sklearn.datasets import make_blobs

# Generate synthetic data
X, y = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=42)

scatter = plt.scatter(X[:, 0], X[:, 1], c=y, cmap="viridis", edgecolor="black")
plt.title("Scatter Plot of Blobs")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.colorbar(label='Cluster')
plt.grid()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "hours": [1, 2, 3, 4, 5, 6],
    "marks": [35, 40, 50, 55, 65, 70]
})
```

```python
plt.scatter(df["hours"], df["marks"], edgecolor="black", color="orange")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Hours vs Marks")
plt.grid()
plt.show()
```

```python
import numpy as np

x = np.arange(1,8)
y = x * 2
data = np.random.rand(100)

fig, axis = plt.subplots(2, 2, figsize=(7,4))

# line plot
axis[0,0].plot(x,y)
axis[0,0].set_title("Line Plot")

# bar plot
axis[0,1].bar(x,y)
axis[0,1].set_title("Bar Plot")

# histogram
axis[1,0].hist(data, bins=20, color="green", edgecolor="black", alpha=0.7)
axis[1,0].set_title("Histogram")

# scatter plot
axis[1,1].scatter(x,y, color="red", edgecolor="black")
axis[1,1].set_title("Scatter Plot")

plt.tight_layout()
plt.grid()
plt.show()
```

```python
# box plot seaborn
import seaborn as sns

data = pd.DataFrame({
    "values": np.random.normal(loc=0, scale=1, size=100),
    "category": np.random.choice(["A", "B", "C"], size=100)
})

sns.boxplot(x="category", y="values", data=data, color="cyan")
plt.title("Box Plot of Normally Distributed Data")
plt.xlabel("Category")
plt.ylabel("Values")
plt.grid()
plt.show()
```

```python
# pair plot

sns.pairplot(data, hue="category")
plt.suptitle("Pair Plot of Categories", y=1.02)
plt.show()
```

```python
# heat map
# data samples
data = pd.DataFrame([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
],
columns=["A", "B", "C"],
index=["Row1", "Row2", "Row3"]
)

heatmap = sns.heatmap(data, annot=True, cmap='coolwarm')
plt.title("Heatmap of Sample Data")
plt.show()
```
