```python
df["Date"] = pd.to_datetime(df["Date"])
df.dtypes
```

```python
df = df.set_index("Date")
print(df)
```

```python
df['year'] = df.index.year
df['month'] = df.index.month
df['day'] = df.index.day
df['dayofweek'] = df.index.day_name()

print(df)
```
